# CRISP-DM — NYC Taxi Trip Duration
A complete lifecycle from business framing through a deployment contract.


## Business understanding
Estimate trip duration using features available at request time. Primary metric: RMSLE; secondary metrics: MAE and median absolute error.

## Data understanding and preparation
Inspect skew, temporal patterns, coordinate validity, and extreme values. Parse timestamps, calculate haversine distance, encode cyclic time, and keep an exclusion audit. Split chronologically before fitting learned transformations.

## Modeling
Compare linear regression with histogram gradient boosting on `log1p(duration)`.

## Evaluation
Compare random and future-month results, report segment errors, and document deployment limitations.

See `../deployment/app.py` for the validated request/response contract prototype.
